```python
import cairo
import math
```

```python
def generate_playing_card(image_path, text, output_path):
    # Set the card dimensions
    width, height = 928, 1312
    
    # Create a Cairo SVG surface and context
    surface = cairo.SVGSurface(output_path, width, height)
    ctx = cairo.Context(surface)
    
    # Load and draw the background image
    background_image = cairo.ImageSurface.create_from_png(image_path)
    ctx.set_source_surface(background_image, 0, 0)
    ctx.paint()
    
    # Set the font and text properties
    ctx.select_font_face("Arial", cairo.FONT_SLANT_NORMAL, cairo.FONT_WEIGHT_BOLD)
    ctx.set_font_size(96)
    
    # Calculate the text position
    text_extents = ctx.text_extents(text)
    x = (width - text_extents.width) / 2
    y = height - 50
    
    # Set the background properties for the text
    bg_width = text_extents.width + 20
    bg_height = text_extents.height + 20
    bg_x = x - 10
    bg_y = y - text_extents.height - 10
    corner_radius = 10
    
    # Draw the background rectangle with rounded corners
    ctx.set_source_rgba(0, 0, 0, 0.7)  # Semi-transparent black background
    ctx.new_sub_path()
    ctx.arc(bg_x + corner_radius, bg_y + corner_radius, corner_radius, 2 * (math.pi / 2), 3 * (math.pi / 2))
    ctx.arc(bg_x + bg_width - corner_radius, bg_y + corner_radius, corner_radius, 3 * (math.pi / 2), 4 * (math.pi / 2))
    ctx.arc(bg_x + bg_width - corner_radius, bg_y + bg_height - corner_radius, corner_radius, 0, math.pi / 2)
    ctx.arc(bg_x + corner_radius, bg_y + bg_height - corner_radius, corner_radius, math.pi / 2, math.pi)
    ctx.close_path()
    ctx.fill()
    
    # Set the text color
    ctx.set_source_rgb(1, 1, 1)  # White text color
    
    # Draw the text
    ctx.move_to(x, y)
    ctx.show_text(text)
    
    # Save the SVG file
    surface.finish()

# Usage example
generate_playing_card("images/questions/ag_frac_2019.png", "Hello, World! lorum ipsum dolores djianndkjnf asnd nasdnk ckdcn", "playing_card.svg")
```
